Problem set 1: Simulating Customer Clicks Using a Bernoulli Random Variable
Problem statement
An organization sends promotional emails. Historical data indicate that a customer
clicks the link with probability 0.3

```python
import numpy as np
import matplotlib.pyplot as plt

# Given probability
p = 0.3
sample_size = 20

# Step 3: Generate Bernoulli observations
samples = np.random.binomial(
    n=1,
    p=p,
    size=sample_size
)

print("Samples:")
print(samples)

# Step 4: Count clicks and non-clicks
number_of_clicks = np.sum(samples == 1)
number_of_non_clicks = np.sum(samples == 0)

print("\nNumber of clicks:", number_of_clicks)
print("Number of non-clicks:", number_of_non_clicks)
print("Total:", number_of_clicks + number_of_non_clicks)

# Step 5: Empirical PMF
empirical_no_click = np.mean(samples == 0)
empirical_click = np.mean(samples == 1)

print("\nEmpirical PMF:")
print("P(X=0):", empirical_no_click)
print("P(X=1):", empirical_click)

# Step 6: Theoretical PMF
theoretical_no_click = 0.7
theoretical_click = 0.3

print("\nTheoretical PMF:")
print("P(X=0):", theoretical_no_click)
print("P(X=1):", theoretical_click)

# Step 8: Theoretical expectation
theoretical_mean = p

# Step 9: Empirical expectation
empirical_mean = np.mean(samples)

# Step 10: Theoretical variance
theoretical_variance = p * (1 - p)

# Step 11: Empirical variance
empirical_variance = np.var(samples)

# Step 12: Final result
print("\nFinal Results")
print("Measure        Theoretical    Empirical    Absolute Difference")

print(
    "Expectation    ",
    theoretical_mean,
    empirical_mean,
    abs(theoretical_mean - empirical_mean)
)

print(
    "Variance       ",
    theoretical_variance,
    empirical_variance,
    abs(theoretical_variance - empirical_variance)
)

# Step 7: Plot
x = np.array([0, 1])
theoretical = [0.7, 0.3]
empirical = [empirical_no_click, empirical_click]

width = 0.35

plt.bar(x - width/2, theoretical, width, label="Theoretical")
plt.bar(x + width/2, empirical, width, label="Empirical")

plt.xlabel("Outcome")
plt.ylabel("Probability")
plt.title("Theoretical vs Empirical PMF")
plt.xticks(x, ["No Click (0)", "Click (1)"])
plt.legend()
plt.show()

# Step 13: Effect of sample size
sample_sizes = [10, 100, 1000, 10000]

print("\nEffect of Sample Size")
print("Sample Size   Empirical Mean   Empirical Variance")

for n in sample_sizes:

    samples_n = np.random.binomial(
        n=1,
        p=p,
        size=n
    )

    mean_n = np.mean(samples_n)
    variance_n = np.var(samples_n)

    print(
        n,
        mean_n,
        variance_n
    )
```

Problem Set 2: ML Use Case: Number of Fraudulent Transactions Flagged
Problem statement: A fraud-detection model examines 50 transactions every
hour. Assume that each transaction has a probability of being flagged as
potentially fraudulent.

```python
import numpy as np

# Given parameters
q = 0.05                 # Replace with the probability in the question
n_transactions = 50
n_hours = 20000

# Step 1: Simulate 20,000 one-hour periods
flagged_transactions = np.random.binomial(
    n=n_transactions,
    p=q,
    size=n_hours
)

# Display first few results
print("First 20 simulated hours:")
print(flagged_transactions[:20])

# Step 2: Number of observations
print("\nNumber of simulated hours:", len(flagged_transactions))

# Step 3: Theoretical expectation
theoretical_mean = n_transactions * q

# Step 4: Empirical expectation
empirical_mean = np.mean(flagged_transactions)

# Step 5: Theoretical variance
theoretical_variance = (
    n_transactions * q * (1 - q)
)

# Step 6: Empirical variance
empirical_variance = np.var(flagged_transactions)

# Step 7: Capacity analysis
hours_over_capacity = np.sum(
    flagged_transactions > 4
)

probability_over_capacity = np.mean(
    flagged_transactions > 4
)

percentage_over_capacity = (
    probability_over_capacity * 100
)

# Display results
print("\nResults")
print("----------------------------------")

print("Theoretical mean:",
      theoretical_mean)

print("Empirical mean:",
      empirical_mean)

print("Theoretical variance:",
      theoretical_variance)

print("Empirical variance:",
      empirical_variance)

print("Hours exceeding capacity:",
      hours_over_capacity)

print("Empirical probability of exceeding capacity:",
      probability_over_capacity)

print("Percentage of hours exceeding capacity:",
      percentage_over_capacity, "%")
```
