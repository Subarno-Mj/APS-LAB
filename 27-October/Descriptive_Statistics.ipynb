```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Create the sample dataset
data = {
    "Student": [f"S{i}" for i in range(1, 16)],
    "Marks": [45, 50, 55, 60, 60, 60, 65, 65,
              70, 70, 75, 75, 80, 85, 90]
}

df = pd.DataFrame(data)
marks = df["Marks"]

# 2. Measures of Central Tendency
mean = marks.mean()
median = marks.median()
mode = marks.mode().tolist()

# 3. Measures of Dispersion
data_range = marks.max() - marks.min()
q1 = np.percentile(marks, 25)
q3 = np.percentile(marks, 75)
iqr = q3 - q1
variance = marks.var(ddof=1)  # Sample variance
std_dev = marks.std(ddof=1)  # Sample standard deviation

# 4. Display the results
print("MEASURES OF CENTRAL TENDENCY")
print(f"Mean: {mean:.2f}")
print(f"Median: {median:.2f}")
print(f"Mode: {mode}")

print("\nMEASURES OF DISPERSION")
print(f"Range: {data_range:.2f}")
print(f"Q1: {q1:.2f}")
print(f"Q3: {q3:.2f}")
print(f"IQR: {iqr:.2f}")
print(f"Sample Variance: {variance:.2f}")
print(f"Sample Standard Deviation: {std_dev:.2f}")

# 5. Create the graphs
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Histogram
sns.histplot(
    data=df, x="Marks", bins=7,
    color="skyblue", edgecolor="black", ax=axes[0, 0]
)
axes[0, 0].set_title("Histogram of Students' Marks")
axes[0, 0].set_xlabel("Marks")
axes[0, 0].set_ylabel("Frequency")

# Scatter plot
sns.scatterplot(
    data=df, x="Student", y="Marks",
    s=100, color="green", ax=axes[0, 1]
)
axes[0, 1].set_title("Scatter Plot of Students' Marks")
axes[0, 1].set_xlabel("Student")
axes[0, 1].set_ylabel("Marks")

# Bar chart
sns.barplot(
    data=df, x="Student", y="Marks",
    color="orange", ax=axes[1, 0]
)
axes[1, 0].set_title("Bar Chart of Students' Marks")
axes[1, 0].set_xlabel("Student")
axes[1, 0].set_ylabel("Marks")

# Box plot
sns.boxplot(
    data=df, x="Marks", color="lightgreen",
    ax=axes[1, 1]
)
axes[1, 1].set_title("Box Plot of Students' Marks")
axes[1, 1].set_xlabel("Marks")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Create the sample dataset
data = {
    "Student": [f"S{i}" for i in range(1, 16)],
    "Marks": [45, 50, 55, 60, 60, 60, 65, 65,
              70, 70, 75, 75, 80, 85, 90]
}

df = pd.DataFrame(data)
marks = df["Marks"]

# 2. Measures of Central Tendency
mean = marks.mean()
median = marks.median()
mode = marks.mode().tolist()

# 3. Measures of Dispersion
data_range = marks.max() - marks.min()
q1 = np.percentile(marks, 25)
q2 = np.percentile(marks, 50)  # Median
q3 = np.percentile(marks, 75)
iqr = q3 - q1
variance = marks.var(ddof=1)
std_dev = marks.std(ddof=1)

# 4. Display the results
print("MEASURES OF CENTRAL TENDENCY")
print(f"Mean: {mean:.2f}")
print(f"Median: {median:.2f}")
print(f"Mode: {mode}")

print("\nMEASURES OF DISPERSION")
print(f"Range: {data_range:.2f}")
print(f"Q1 (First Quartile): {q1:.2f}")
print(f"Q2 (Second Quartile / Median): {q2:.2f}")
print(f"Q3 (Third Quartile): {q3:.2f}")
print(f"IQR: {iqr:.2f}")
print(f"Sample Variance: {variance:.2f}")
print(f"Sample Standard Deviation: {std_dev:.2f}")

# 5. Create the graphs
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Histogram
sns.histplot(
    data=df, x="Marks", bins=7,
    color="skyblue", edgecolor="black", ax=axes[0, 0]
)
axes[0, 0].set_title("Histogram of Students' Marks")
axes[0, 0].set_xlabel("Marks")
axes[0, 0].set_ylabel("Frequency")

# Scatter plot
sns.scatterplot(
    data=df, x="Student", y="Marks",
    s=100, color="green", ax=axes[0, 1]
)
axes[0, 1].set_title("Scatter Plot of Students' Marks")
axes[0, 1].set_xlabel("Student")
axes[0, 1].set_ylabel("Marks")

# Bar chart
sns.barplot(
    data=df, x="Student", y="Marks",
    color="orange", ax=axes[1, 0]
)
axes[1, 0].set_title("Bar Chart of Students' Marks")
axes[1, 0].set_xlabel("Student")
axes[1, 0].set_ylabel("Marks")

# Box plot with Q1, Q2, Q3 labels
sns.boxplot(
    data=df, x="Marks",
    color="lightgreen", ax=axes[1, 1]
)

# Mark Q1, Q2, and Q3
ax = axes[1, 1]
ax.axvline(q1, color="blue", linestyle="--", linewidth=2)
ax.axvline(q2, color="red", linestyle="--", linewidth=2)
ax.axvline(q3, color="purple", linestyle="--", linewidth=2)

# Add labels above the box plot
y_position = 0.25
ax.text(q1, y_position, f"Q1 = {q1:.0f}",
        color="blue", ha="center", va="bottom", fontweight="bold")
ax.text(q2, y_position, f"Q2 = {q2:.0f}",
        color="red", ha="center", va="bottom", fontweight="bold")
ax.text(q3, y_position, f"Q3 = {q3:.0f}",
        color="purple", ha="center", va="bottom", fontweight="bold")

ax.set_title("Box Plot with Quartiles (Q1, Q2, Q3)")
ax.set_xlabel("Marks")
ax.set_yticks([])

plt.tight_layout()
plt.show()
```
